## Pruebas paramétricas

### Prueba t para 1 muestra
**Prueba de hipótesis de 2 colas (bilateral)**

**Problema:** Determinar si la media de los puntajes es diferente de 0.067.
- **Hipótesis nula (H0):** La media de los puntajes no es diferente a 0.067.
- **Hipótesis alternativa (H1):** La media de los puntajes es diferente a 0.067.

```python
import numpy as np
from scipy import stats

np.random.seed(1)
puntajes = np.random.normal(loc=0, scale=1, size=1000)
media_hipotetica = 0.067

t_stat, p_value = stats.ttest_1samp(puntajes, media_hipotetica)
print(f'El valor de t es: {np.round(t_stat, 5)}')
print('='*50)
print(f'El valor p es: {np.round(p_value, 5)}')
print('='*50)
alpha = 0.05
if p_value < alpha:
    print('Se rechaza la hipótesis nula')
else:
    print('No se rechaza la hipótesis nula')
```

### Prueba Anova para mas de 2 muestras independientes (anova factorial)

**Problema:** Determinar si existe diferencia entre la media de los puntajes de los grupos A, B y C.
- **Hipótesis nula (H0):** La media de los puntajes de los grupos A, B y C no son diferentes.
- **Hipótesis alternativa (H1):** Al menos una de las medias de los puntajes de los grupos A, B y C es diferente a las demás.

```python
import numpy as np
import pandas as pd
from scipy.stats import f_oneway
from statsmodels.stats.multicomp import pairwise_tukeyhsd
np.random.seed(4)
grupo_a = np.random.normal(loc=0, scale=1, size=100)
grupo_b = np.random.normal(loc=0, scale=1, size=100)
grupo_c = np.random.normal(loc=0, scale=1, size=100)
grupos = pd.DataFrame({'A': grupo_a, 'B': grupo_b, 'C': grupo_c})
print(grupos.head())
```

```python
grupos_melt = pd.melt(grupos.reset_index(), id_vars=['index'], value_vars=['A','B', 'C'],
                     var_name="Grupo", value_name="Puntaje").rename(columns={"index":"Sujeto"})
print(grupos_melt.head())
```

```python
f_stat, p_value = f_oneway(grupo_a, grupo_b, grupo_c)
print(f'El valor de f es: {np.round(f_stat, 5)}')
print('='*50)
print(f'El valor p es: {np.round(p_value, 5)}')
print('='*50)
alpha = 0.05
if p_value < alpha:
    print('\nSe rechaza la hipótesis nula\n')
else:
    print('\nNo se rechaza la hipótesis nula\n')
```

```python
# aplicando la prueba post hoc de Tukey
tukey = pairwise_tukeyhsd(endog=grupos_melt['Puntaje'],
                          groups=grupos_melt['Grupo'], alpha=0.05)
print(tukey)
```

### Anova para mas de 2 muestras dependientes (ANOVA de una vía de medidas repetidas)

**Problema:** Determinar si hay diferencias en la media de los puntajes de los sujetos en las pruebas A, B y C.
- **Hipótesis nula (H0):** La media de los puntajes de las pruebas A, B y C en el mismo grupo
de sujetos no son diferentes.
- **Hipótesis alternativa (H1):** Al menos una de las medias de los puntajes de las pruebas
A, B o C en el mismo grupo de sujetos es diferente.

```python
import pandas as pd
import numpy as np
from statsmodels.stats.anova import AnovaRM
from statsmodels.stats.multicomp import pairwise_tukeyhsd
np.random.seed(5)
sujetos = np.arange(0, 100)
prueba_a = np.random.normal(loc=0, scale=1, size=100)
prueba_b = np.random.normal(loc=0, scale=1, size=100)
prueba_c = np.random.normal(loc=0, scale=1, size=100)
pruebas = pd.DataFrame({'Sujetos': sujetos, 'Prueba_a': prueba_a, 'Prueba_b': prueba_b, 'Prueba_c': prueba_c})
print(pruebas.head())
```

```python
pruebas_melt = pd.melt(pruebas, id_vars=['Sujetos'], var_name='Prueba', value_name='Puntaje')
print(pruebas_melt.head())
```

```python
anova_rm = AnovaRM(pruebas_melt, 'Puntaje', 'Sujetos', within=['Prueba'])
resultados = anova_rm.fit()
print(resultados)
valor_p = resultados.anova_table['Pr > F'].iloc[0]
alpha = 0.05
if valor_p < alpha:
    print('\nSe rechaza la hipótesis nula\n')
else:
    print('\nNo se rechaza la hipótesis nula\n')
```

```python
# aplicando la prueba post hoc de Tukey
tukey = pairwise_tukeyhsd(endog=pruebas_melt['Puntaje'],
                          groups=pruebas_melt['Prueba'], alpha=0.05)
print('='*50)
print(tukey)
```

### Correlación de Pearson

**Problema:** Determinar si existe una relación lineal entre las variables X e Y.
- **Hipótesis Nula (H0):** La correlación de Pearson entre las variables X e Y es igual a 0, indicando que no hay una relación lineal entre ellas.
- **Hipótesis Alternativa (H1):** La correlación de Pearson entre las variables X e Y es diferente de 0, indicando que existe una relación lineal entre ellas.

```python
import numpy as np
import pandas as pd
from scipy.stats import pearsonr
np.random.seed(6)
datos_1 = np.random.normal(loc=0, scale=1, size=100)
datos_2 = np.random.normal(loc=0, scale=1, size=100)
df = pd.DataFrame({'X': datos_1, 'Y': datos_2})
print(df.head())
```

```python
correlation, p_value = pearsonr(df['X'], df['Y'])
print('El valor de la correlación es: %f' %(correlation))
print('='*50)
print('El valor del p valor es: %f' %(p_value))
print('='*50)
alpha = 0.05
if p_value < alpha:
    print('Se rechaza la hipótesis nula')
else:
    print('No se rechaza la hipótesis nula')
```

**Resumen**

- **Valor p < 0.05:** Correlación significativa (rechazar H0).
- **Valor p > 0.05:** Correlación no significativa (no rechazar H0).

**Correlación de 1 variable entre diferentes variables**

```python
import numpy as np
import pandas as pd
from scipy.stats import pearsonr
np.random.seed(6)
variable_w = np.random.normal(loc=0, scale=1, size=100)
variable_x = np.random.normal(loc=0, scale=1, size=100)
variable_y = np.random.normal(loc=0, scale=1, size=100)
variable_z = np.random.normal(loc=0, scale=1, size=100)
variables = [variable_w, variable_x, variable_y, variable_z]
resultados = {}
letter = ['w','x','y','z']
for l, var in zip(letter, variables):
    correlation, pvalue = pearsonr(variable_w, var)
    resultados[f'Variable_{l}'] = {'correlation': correlation,
                                   'p_value': p_value}
for var, stats in resultados.items():
    print('='*50)
    print(f"Correlación entre la variable_w y {var}: {stats['correlation']:.6f}")
    print(f"valor p entre la variable_w y {var}: {stats['p_value']:.6f}")
```

**Problema:** Determinar si la mediana de los puntajes después de la intervención es menor que la mediana de los puntajes antes de la intervención en el mismo grupo de sujetos.
- **Hipótesis nula (H0):** La mediana de los puntajes después de la intervención no es igual o menor que la mediana de los puntajes antes de la intervención en el mismo grupo de sujetos.
- **Hipótesis alternativa (H1):** La mediana de los puntajes después de la intervención es menor que la mediana de los puntajes antes de la intervención en el mismo grupo de sujetos.

```python
import numpy as np
from scipy.stats import wilcoxon
np.random.seed(8)
antes = np.random.randint(1, 10, 100)
despues = np.random.randint(1, 10, 100)
# Aplicando la prueba de Wilcoxon de una cola (inferior)
w_stat, p_value = wilcoxon(antes, despues, alternative='less')
print(f'El valor de t es: {w_stat}')
print('='*50)
print(f'El valor p es: {np.round(p_value, 5)}')
print('='*50)
alpha = 0.05
if p_value < alpha:
    print('Se rechaza la hipótesis nula')
else:
    print('No se rechaza la hipótesis nula')
```

### Prueba kruskal-wallis para mas de 2 muestras independientes

**Problema:** Determinar si hay una diferencia entre la mediana de los puntajes de los grupos A, B y C.
- **Hipótesis nula (H0):** La mediana de los puntajes de los grupos A, B y C no son diferentes.
- **Hipótesis alternativa (H1):** Al menos uno de las medianas de los puntajes de los grupos A, B o C es diferente.

```python
import pandas as pd
from scipy.stats import kruskal
import numpy as np

np.random.seed(10)
grupo_a = np.random.randint(0, 10, 1000)
grupo_b = np.random.randint(0, 10, 1000)
grupo_c = np.random.randint(0, 10, 1000)
grupos = pd.DataFrame({'A': grupo_a, 'B': grupo_b, 'C': grupo_c})
print(grupos.head())
```

```python
grupos_melt = pd.melt(grupos.reset_index(), id_vars=['index'],
                      value_vars=['A', 'B', 'C'], var_name="Grupos").rename(columns={"index":"Sujetos"})
print(grupos_melt.head())
```

```python
h_stat, p_value = kruskal(grupo_a, grupo_b, grupo_c)
print(f'El valor de t es: {np.round(h_stat, 5)}')
print('='*50)
print(f'El valor p es: {np.round(p_value, 5)}')
print('='*50)
alpha = 0.05
if p_value < alpha:
    print('Se rechaza la hipótesis nula')
else:
    print('No se rechaza la hipótesis nula')
```

### Correlacion entre 2 variables (spearman)

**Problema:** Determinar si existe una relación monótona entre las variables X e Y.
- **Hipótesis nula (H0):** La correlación de Spearman entre las variables X e Y es igual a 0, indicando que no hay una relación monótona entre ellas.
- **Hipótesis alternativa (H1):** La correlación de Spearman entre las variables X e Y es
diferente de 0, indicando que existe una relación monótona entre ellas.

```python
import numpy as np
from scipy.stats import spearmanr
np.random.seed(12)
variable_x = np.random.randint(0, 10, 100)
variable_y = np.random.randint(0, 10, 100)
correlation, p_value = spearmanr(variable_x, variable_y)
print(f'El valor de la correlación es: {np.round(correlation, 5)}')
print('='*50)
print(f'El valor p es: {np.round(p_value, 5)}')
alpha = 0.05
print('='*50)
if p_value < alpha:
    print('Se rechaza la hipótesis nula')
else:
    print('No se rechaza la hipótesis nula')
```

**Correlación de 1 variable entre diferentes variables**

```python
import numpy as np
from scipy.stats import spearmanr
np.random.seed(12)
variable_w = np.random.randint(0, 10, 100)
variable_x = np.random.randint(0, 10, 100)
variable_y = np.random.randint(0, 10, 100)
variable_z = np.random.randint(0, 10, 100)
variables = [variable_w, variable_x, variable_y, variable_z]

letters = ['w', 'x', 'y', 'z']
resultados = {}
for i, var in zip(letters, variables):
    correlation, p_value = spearmanr(variable_w, var)
    resultados[f'variable_{i}'] = {'correlation': correlation, 'p_value': p_value}
# imprimir resultados
for var, stats in resultados.items():
    print('='*50)
    print(f'Correlacion entre la variable_w y la {var}: {np.round(stats["correlation"], 5)}')
    print(f'Valor p entre la variable_w y {var}: {np.round(stats["p_value"],5)}')
```
