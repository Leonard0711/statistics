## Medidas de tendencia central
### Cálculo de la media

```python
def media(valores: list[float]) -> float:
    return sum(valores) / len(valores)

# print(media([2,3,4,5,6]))
```

### Cálculo de la mediana

```python
def mediana_impar(valores: list[float]) -> float:
    return sorted(valores)[len(valores)//2]

# print(mediana_impar([2,3,4,5,6]))
```

```python
def mediana_par(valores: list[float]) -> float:
    ordenado = sorted(valores)
    punto_medio = len(valores) // 2
    return (ordenado[punto_medio - 1] + ordenado[punto_medio]) / 2

# print(mediana_par([2,3,4,5,7,6]))
```

```python
def calcular_mediana(valores: list[float]) -> float:
    return mediana_par(valores) if len(valores) % 2 == 0 else mediana_impar(valores)

# print(calcular_mediana([2,4,5,6,7]))
```

### Cálculo de la moda

```python
def calcular_moda(valores: list[float]) -> float:
    conteos = {}
    for i in valores:
        if i not in conteos:
            conteos[i] = 0
        conteos[i] += 1
    return max(conteos, key=conteos.get)

# print(calcular_moda([1,1,2,3,3,3,4,5,6,6,7]))       
```
