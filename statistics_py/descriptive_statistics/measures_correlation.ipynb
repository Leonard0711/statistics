### Cálculo de covarianza

```python
def sum_producto_interno(x: list[float], y: list[float]) -> float:
    assert len(x) == len(y), "el tamaño de ambos listas deben ser iguales"
    return sum(x_i * y_i for x_i, y_i in zip(x, y))

print(sum_producto_interno([1,2,3], [3,2,1]))
```

```python
def covarianza(x: list[float], y: list[float]) -> float:
    assert len(x) == len(y), "el tamaño de las variables deben ser iguales"
    dif_medias_x = diferencia_media(x)
    dif_medias_y = diferencia_media(y)
    suma_producto = sum_producto_interno(dif_medias_x, dif_medias_y)
    return suma_producto / (len(x) - 1)

print(covarianza([1,2,3,4,5,6,7,8], [8,7,6,5,4,3,2,1]))
```

### Cálculo de correlación

```python
def correlacion_pearson(x: list[float], y: list[float]) -> float:
    assert len(x) == len(y), "el tamaño de las variables deben ser iguales"
    cov_x_y = covarianza(x, y)
    return cov_x_y / (desviacion_estandar(x)*desviacion_estandar(y))

print(correlacion_pearson([1,2,3,4,5,6,7,8], [8,7,6,5,4,3,2,1]))
```
