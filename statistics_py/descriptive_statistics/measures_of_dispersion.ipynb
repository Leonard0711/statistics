## Medidas de dispersión

### Cálculo de varianza

```python
def diferencia_media(valores: list[float]) -> list[float]:
    valor_media = media(valores)
    return [valor - valor_media for valor in valores]

# print(diferencia_media([1,2,3,4,5,6]))
```

```python
def suma_cuadrados(valores: list[float]) -> float:
    total = 0
    for i in valores:
        total += i**2
    return total

# print(suma_cuadrados([1,2,3]))
```

```python
def varianza_muestral(valores: list[float]) -> float:
    diferencias = diferencia_media(valores)
    suma_cuadrados_ = suma_cuadrados(diferencias)
    return suma_cuadrados_ / (len(valores) - 1)

# print(varianza_muestral([1,2,3,4,5,6,7,8,8]))
```

### Cálculo de desviación estándar

```python
import math

def desviacion_estandar(valores: list[float]) -> float:
    return math.sqrt(varianza_muestral(valores))

# print(desviacion_estandar([1,2,3,4,5,6,7,8,8]))
```
