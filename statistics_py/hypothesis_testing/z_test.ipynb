## Importaciones

````python
import numpy as np
from scipy.stats import norm
from statsmodels.stats.weightstats import ztest
````

## Ejemplo de datos:

````python
datos = np.array([102, 98, 101, 105, 99,
                  103, 100, 97, 104, 101])
````

## Hipótesis

Supongamos:

$$ H_0:\mu=100 $$ $$ H_1:\mu\neq100 $$

### Estadístico z manual

````python
mu_0 = 100
mu_sample = np.mean(datos)
sigma = 5
n = len(datos)

z_calculated = (mu_sample - mu_0) / (sigma / np.sqrt(n))

print(z_calculated)
````

Se muestra que para el z-test clásico de una media se utiliza la desviación estándar poblacional $\sigma$.

### Valor crítico

````python
alpha = 0.05

z_critical = norm.ppf(1 - alpha / 2)

print(z_critical)
````

### p-value

````python
p_value = 2 * norm.sf(abs(z_calculated))

print(p_value)
````

## Con statsmodels
- ```ztest``` estima la desviación estándar a partir de los datos muestrales no utiliza el ```sigma``` designado. 

````python
z_stat, p_value = ztest(datos, value=100)

print("z:", z_stat)
print("p-value:", p_value)
````

y finalmente:

````python
if p_value < alpha:
    print("Se rechaza H0")
else:
    print("No se rechaza H0")
````
