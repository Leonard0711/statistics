## Importaciones

```python
import numpy as np
from scipy.stats import ttest_1samp
from scipy.stats import t
```

## Datos

```python
datos = np.array([102, 98, 101, 105, 99,
                  103, 100, 97, 104, 101])
```

## t calculado manualmente
$$ t= \frac{\bar{x}-\mu_0} {s/\sqrt{n}} $$

```python
mu_0 = 100

mean = np.mean(datos)
s = np.std(datos, ddof=1)
n = len(datos)

t_calculated = (mean - mu_0) / (s / np.sqrt(n))

print(t_calculated)
```

## Grados de libertad

```python
df = n - 1

print(df)
```

Aquí queda claro:

$$ df=n-1 $$

## t crítico

```python
alpha = 0.05

t_critical = t.ppf(1 - alpha / 2, df=df)

print(t_critical)
```

## p-value

```python
p_value = 2 * t.sf(abs(t_calculated), df=df)
print(p_value)
```

## Con SciPy

```python
t_stat, p_value = ttest_1samp(datos, popmean=100)

print("t:", t_stat)
print("p-value", p_value)
```
