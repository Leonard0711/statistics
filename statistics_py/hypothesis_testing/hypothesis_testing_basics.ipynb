## La hipótesis nula y alternativa

````python
H0 = "La media poblacional es igual a 100"
H1 = "La media poblacional es diferente de 100"
````

Conceptualmente:

$$ H_0:\mu=100 $$ $$ H_1:\mu\neq100 $$

## Nivel de significancia

````python
alpha = 0.05
````

Explica que:

$$ \alpha=P(\text{rechazar }H_0\mid H_0\text{ verdadera}) $$

### p-value

Ejemplo:

````python
p_value = 0.03

if p_value < alpha:
    print("Se rechaza H0")
else:
    print("No se rechaza H0")
````

Regla:

$$ p<\alpha \Rightarrow \text{rechazar }H_0 $$

### Prueba de una cola y dos colas
```Una cola izquierda:```

```Una cola derecha:```

```H1: μ > μ0```


```Dos colas:```

```H1: μ ≠ μ0```
