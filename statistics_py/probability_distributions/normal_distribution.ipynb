## Importación

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
````

## Parámetros de la distribución normal

La distribución normal se define principalmente mediante:

* ```loc``` → media μ
* ```scale``` → desviación estándar σ

````python
mu = 100
sigma =15
````

## PDF — función de densidad

* ```pdf()``` devuelve la densidad de probabilidad para un valor ```x```.

````python
x = 100
density = norm.pdf(x, loc=mu, scale=sigma)
print(density)
````

Para varios valores:

````python
x = np.linspace(40, 160, 500)
pdf = norm.pdf(x, loc=mu, scale=sigma)
````

Visualización

````python
plt.plot(x, pdf)
plt.xlabel("X")
plt.ylabel("Densidad")
plt.title("Distribución normal")
plt.show()
````

## CDF - probabilidad acumulada

```cdf()``` responde:

- ¿Cuál es la probabilidad de obtener un valor menor o igual que X?

Por ejemplo:

````python
x = 120
probabilidad = norm.cdf(x, loc=mu, scale=sigma)
print(probabilidad)
````

Matemáticamente:

$$ P(X \leq 120) $$

También puedes calcular probabilidades entre dos valores:

````python
p = norm.cdf(120, loc=mu, scale=sigma) - norm.cdf(80, loc=mu, scale=sigma)
print(p)
````

Esto representa:

$$ P(80 \leq X \leq 120) $$

## PPF - percentiles
```ppf()``` realiza el proceso inverso de ```cdf()```.

Si conoces una probabilidad y quieres encontrar el valor de ```X```:

````python
p = 0.95
x = norm.ppf(p, loc=mu, scale=sigma)
print(x)
````

Esto responde:

- ¿Qué valor de X deja al 95 % de las observaciones por debajo?

Es decir:

$$ P(X \leq x)=0.95 $$

## Relación entre CDF y PPF
Esta parte es importante para consolidar lo visto anteriormente:

````python
x = norm.ppf(0.95, loc=mu, scale=sigma)

p = norm.cdf(x, loc=mu, scale=sigma)

print(x)
print(p)
````

La relación es: \
```cdf```:\
X ───────────────► Probabilidad

```PPF```:\
Probabilidad ────► X

## Z - score
El z-score estandariza un valor:

$$ z = \frac{x-\mu}{\sigma} $$

En Python:

````python
x = 120
z = (x - mu) / sigma
print(z)
````

Es decir, 120 está aproximadamente 1.33 desviaciones estándar por encima de la media.

También se puede obtener la probabilidad utilizando directamente la normal estándar:

````python
probabilidad = norm.cdf(z)

print(probabilidad)
````

## Comparando el normal general y normal estándar

````python
x = 120
z = (x - mu) / sigma

p1 = norm.cdf(x, loc=mu, scale=sigma)
p2 = norm.cdf(z)

print(p1)
print(p2)
````

Esto permite demostrar la transformación de cualquier variable normal en una normal estándar y viceversa.:

$$ X \sim N(\mu,\sigma) $$

y

$$ Z \sim N(0,1) $$


## Variables críticos
Por ejemplo, encontrar el valor correspondiente al 97.5 %:

````python
z_critical = norm.ppf(0.975)

print(z_critical)
````

Esto permite conectar la distribución normal con **intervalos de confianza y pruebas de hipótesis**.

Para una prueba bilateral:

````python
alpha = 0.05

z_critical = norm.ppf(1 - alpha / 2)

print(z_critical)
````

## Probabilidades de una sola cola
* Cola derecha:

````python
x = 130

p_right = 1 - norm.cdf(x, loc=mu, scale=sigma)

print(p_right)
````

* Cola izquierda:

````python
p_left = norm.cdf(x, loc=mu, scale=sigma)

print(p_left)
````
